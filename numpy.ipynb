**Valid Sudoku**

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
target = np.arange(1, 10)

# har row aur column ko sort karo, 1 se 9 ban jaana chahiye
rows_ok = np.array_equal(np.sort(s, axis=1), np.tile(target, (9, 1)))
cols_ok = np.array_equal(np.sort(s, axis=0), np.tile(target.reshape(9, 1), (1, 9)))

# 3x3 boxes ko 9 rows mein badlo (har row = ek box)
boxes = s.reshape(3, 3, 3, 3).swapaxes(1, 2).reshape(9, 9)
boxes_ok = np.array_equal(np.sort(boxes, axis=1), np.tile(target, (9, 1)))

print("Rows valid :", rows_ok)
print("Cols valid :", cols_ok)
print("Boxes valid:", boxes_ok)
print("Valid Sudoku?", rows_ok and cols_ok and boxes_ok)
```

**General Questions**

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
print("Shape:", data.shape)
```

```python
#Find the average age of students.
print("Average age:", np.mean(data[:, 0]))
```

```python
#Extract Math marks of all students.
print("Math marks:", data.T[1])
```

```python
#Find the highest Science mark.
print("Highest Science mark:", np.amax(data[:, 2]))
```

```python
#Get details of the student who scored more than 90 in Math.
idx = np.where(data[:, 1] > 90)[0]
print("Student number:", idx + 1)
print("[Age, Math, Science]:", data[idx])
```

```python
#Increase Math marks of all students by 5.
# copy par kaam kiya, taaki neeche ke sawaalon mein original data hi use ho
bumped = data.copy()
bumped[:, 1] = bumped[:, 1] + 5
print("New Math marks:", bumped[:, 1])
```

```python
#Find how many students are younger than 19.
print("Students younger than 19:", np.count_nonzero(data[:, 0] < 19))
```

```python
#Calculate the average marks in each subject (column-wise mean).
# Age column (0) chhod ke sirf Math aur Science
print("[Math avg, Science avg]:", np.mean(data[:, 1:], axis=0))
```

```python
#Get data of students who scored at least 80 in both subjects.
both = np.all(data[:, 1:] >= 80, axis=1)
print(data[both])
```

```python
#Replace all Science marks < 75 with 0.
result = data.copy()
result[result[:, 2] < 75, 2] = 0
print(result)
```
